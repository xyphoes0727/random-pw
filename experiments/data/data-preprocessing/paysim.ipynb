```python
import numpy as np
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print(os.getcwd())
df = pd.read_csv(f"../datasets/paysim.csv")
```

```python
df = df.drop(columns=["Unnamed: 0"], axis=1)
```

```python
df.head()
```

```python
histCols = ["oldbalanceOrg", "oldbalanceDest", "newbalanceOrig", "newbalanceDest"]
toDropHist = [col for col in df.columns if col not in histCols]
print(toDropHist)
dfToHist = df.drop(columns=toDropHist , axis=1)
```

```python
for col in histCols:
    df[col] = np.log1p(df[col])
```

```python
sns.histplot(data=df[["newbalanceOrig"]], x= "newbalanceOrig", bins=25)
plt.show()
```

```python
sns.histplot(data=df[["oldbalanceOrg"]], x= "oldbalanceOrg", bins=25)
plt.show()
```

```python
sns.histplot(data=df[["newbalanceDest"]], x= "newbalanceDest", bins=25)
plt.show()
```

```python
sns.histplot(data=df[["oldbalanceDest"]], x= "oldbalanceDest", bins=25)
plt.show()
```

```python
df.to_csv(f"../datasets/paysim-cleaned.csv", index=False)
```
